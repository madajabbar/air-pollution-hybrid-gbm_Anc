# Eksperimen 3: Tahap-2 (Koreksi Residual)

Dengan **LightGBM** tahap-1, membandingkan **BiLSTM**, **Theta**, dan **TFT** sebagai tahap-2 (koreksi residual).

```
import pandas as pd, re, sys
BASE='/home/mada/Documents/Kuliah/Algorithms and Complexity/02_Kode/hybrid-lgbm-xgb'
sys.path.insert(0,BASE)
lines=[l for l in open(f'{BASE}/hasil_fair_full.txt') if '[OK]' in l]
data=[]
for l in lines:
    m=re.match(r'\[OK\] (\S+) h(\d+) \+(\S+): RMSE=([\d.]+)',l)
    if m: data.append(dict(pol=m.group(1),hz=int(m.group(2)),s2=m.group(3),rmse=float(m.group(4))))
df=pd.DataFrame(data)
print(df.pivot_table(index='s2',columns=['pol','hz'],values='rmse'))
```

```
import matplotlib.pyplot as plt
fig,axes=plt.subplots(1,3,figsize=(13,4),sharey=True)
for ax,p in zip(axes,['PM25','PM10','NO2']):
    d=df[df.pol==p].pivot(index='s2',columns='hz',values='rmse')
    d.plot(kind='bar',ax=ax,title=p,legend=False)
    ax.grid(alpha=0.3)
axes[0].set_ylabel('RMSE'); axes[2].legend(loc='best'); fig.suptitle('Tahap-2: BiLSTM vs Theta vs TFT')
plt.tight_layout(); plt.show()
```
