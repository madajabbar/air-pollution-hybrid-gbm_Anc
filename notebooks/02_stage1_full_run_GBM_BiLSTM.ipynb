# Eksperimen 2: Full Run GBM + BiLSTM

Menjalankan **LightGBM** dan **XGBoost** (tahap-1) dengan **BiLSTM** (tahap-2) pada semua 3 polutan × 3 horizon = 18 kombinasi.

```
import pandas as pd, sys
BASE='/home/mada/Documents/Kuliah/Algorithms and Complexity/02_Kode/hybrid-lgbm-xgb'
sys.path.insert(0,BASE)
lines=[l for l in open(f'{BASE}/hasil_full_bilstm.txt') if '[OK]' in l]
import re
data=[]
for l in lines:
    m=re.match(r'\[OK\] (\S+)\+bilstm (\S+) h(\d+): RMSE=([\d.]+) MAE=([\d.]+) MAPE=([\d.]+)%',l)
    if m: data.append(dict(stage1=m.group(1),pol=m.group(2),hz=int(m.group(3)),
                          rmse=float(m.group(4)),mae=float(m.group(5)),mape=float(m.group(6))))
df=pd.DataFrame(data)
print(df.pivot_table(index='stage1',columns=['pol','hz'],values='rmse'))
```

```
import matplotlib.pyplot as plt
fig,axes=plt.subplots(1,3,figsize=(13,4),sharey=True)
for ax,p in zip(axes,['PM25','PM10','NO2']):
    d=df[df.pol==p]
    for s1,c in [('lightgbm','orange'),('xgboost','green')]:
        sub=d[d.stage1==s1].set_index('hz').rmse.sort_index()
        ax.plot(sub.index.astype(str),sub.values,marker='o',label=s1,color=c)
    ax.set_title(p); ax.grid(alpha=0.3)
axes[0].set_ylabel('RMSE'); axes[2].legend(); fig.suptitle('Full run GBM+BiLSTM')
plt.tight_layout(); plt.show()
```
